# Track 1, Step 1: Phenotype Extraction (HPO Mapping)
**Goal**: Read the clinical document and extract standardized medical symptoms (HPO terms).
**Input**: `data/Challenge_Clinical_Phenotype_1.docx`


```python
import docx
import pandas as pd

# Load the clinical document
doc = docx.Document('../data/Challenge_Clinical_Phenotype_1.docx')

# The HPO terms are conveniently provided in a table within the document.
# Let's extract them into a pandas DataFrame.
data = []
for table in doc.tables:
    for i, row in enumerate(table.rows):
        if i == 0:
            headers = [cell.text.strip().replace('\n', ' ') for cell in row.cells]
            continue
        row_data = [cell.text.strip().replace('\n', ' ') for cell in row.cells]
        data.append(row_data)

hpo_df = pd.DataFrame(data, columns=headers)
hpo_df
```

### Extracted HPO Terms
We will use these terms later to cross-reference with the gene mutations we find in the VCF file. 
Let's store them in a simple list.

```python
hpo_terms = hpo_df['HPO ID'].tolist()
print(f"Extracted {len(hpo_terms)} HPO terms:")
print(hpo_terms)
```
